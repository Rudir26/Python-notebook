```python
#tehtava 2
number1 = 7; number2 = 8; number3 = 7

#a
if number1 == number2 == number3:
    print("All numbers are equal")
else:
    print("numbers are not equal")

#b
if number1 == number2:
    print("number1 and number2 are equal")
elif number2 == number3:
    print("number2 and number3 are equal")
else:
    print("neither number1 and number2 nor number2 and number3 are equal")

#c
if number1 > number2:
    print("number1 is greater than number2")
else:
    print("number1 is not greater than number2")

if number1 > number3:
    print("number1 is greater than number3")
else:
    print("number1 is not greater than number3")

#d
if number1 > number2:
    print("number1 is greater than number2")
else:
    if number1 > number3:
        print("number1 is greater than number3")

#e
if number1 == number2:
    print("numbers 1 and 2 are equal")
else:
    if number1 == number3:
        print("numbers 1 and 3 are equal")
    else:
        print("none of the numbers are equal")
```
